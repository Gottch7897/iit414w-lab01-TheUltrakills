```python
from pathlib import Path

import pandas as pd
import yfinance as yf

symbols = {
    "Microsoft": "MSFT",
    "Apple": "AAPL",
    "Meta": "META",
    "Oracle": "ORCL",
    "Alphabet": "GOOGL",
    "Amazon": "AMZN",
    "NVIDIA": "NVDA",
    "Adobe": "ADBE",
    "Salesforce": "CRM",
    "Intel": "INTC",
    "S&P 500": "^GSPC",
}

output_file = Path.cwd() / "ejercicio4.csv"
end_date = pd.Timestamp.today().normalize() + pd.Timedelta(days=1)
start_date = end_date - pd.DateOffset(years=5)
price_series = []

for company, ticker in symbols.items():
    data = yf.download(
        ticker,
        start=start_date,
        end=end_date,
        auto_adjust=False,
        progress=False,
    )
    if data.empty:
        print(f"Se omitio {company} ({ticker}) porque no devolvio datos.")
        continue

    series = data[["Adj Close"]].rename(columns={"Adj Close": ticker})
    series.index.name = "Date"
    price_series.append(series)

if not price_series:
    raise RuntimeError("No se pudieron descargar datos para ninguno de los tickers solicitados.")

wide = pd.concat(price_series, axis=1, join="inner").dropna()
wide = wide.reset_index()
wide.to_csv(output_file, index=False)

print(f"Rango comun: {start_date.date()} a {end_date.date()}")
print(f"CSV creado: {output_file}")
print(f"Empresas incluidas: {len(price_series)}")
print(f"Filas totales: {len(wide):,}")
print(f"Columnas: {list(wide.columns)}")
```
